# Target vs Features

```python
y = df['HadHeartAttack'].replace({"No": 0, "Yes": 1})

corr = df.select_dtypes("number").join(y).corr()
corr
```

```python
sns.heatmap(corr, cmap="coolwarm", annot=True)
```

```python
rates = (
    df.assign(HeartAttack=df["HadHeartAttack"].eq("Yes"))
      .groupby("AgeCategory", sort=True)
      .agg(n=("HeartAttack", "size"),
           heart_attack_rate=("HeartAttack", "mean"))
)
rates["heart_attack_pct"] = 100 * rates["heart_attack_rate"]
display(rates[["n", "heart_attack_pct"]])
rates["heart_attack_pct"].plot.bar(
    ylabel="Reported heart attack (%)",
    title="Heart attack history by age group"
)
```

```python
health = (df.assign(HeartAttack=df["HadHeartAttack"].eq("Yes"))
        .groupby("GeneralHealth")
        .agg(
            n=("HeartAttack", "size"), 
            heart_attack_rate=("HeartAttack", "mean"))
)

health["heart_attack_pct"] = 100 * health["heart_attack_rate"]

health = health.sort_values('heart_attack_pct', ascending=True)

display(health[["n", "heart_attack_pct"]])
health["heart_attack_pct"].plot.bar(
    ylabel="Reported heart attack (%)",
    title="Heart attack history general health"
)
```

Findings
 * BMI and weight have strong correlation due to their mathematical dependence.
 * Physical Health Days has the strongest correlation with HadHeartAttack (0.13). (numerical featuers only)
 * Higher age ranges have greater reported heart attacks. 13.6% for 80+ age, 0.38% for 18-24 age.
 * Poor health results in higher reported heart attacks (21.45%) vs excellent health (1.43%).
